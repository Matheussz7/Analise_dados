```python
planilha_bruta = pd.read_excel(caminho_arquivo, sheet_name="Municípios", header=None)
```

```python
base_demografica = planilha_bruta.iloc[3:5573].copy()

```

```python
base_demografica
```

```python
base_demografica.columns = [
    "coluna_auxiliar",
    "sigla_uf",
    "codigo_uf",
    "codigo_mun",
    "nome_municipio",
    "pop_2010_base",
    "pop_2010_ajustada",
    "pop_2022",
]

base_demografica = base_demografica.drop(columns=["coluna_auxiliar"])
base_demografica = base_demografica.reset_index(drop=True)
base_demografica
```

```python
for col in ["pop_2010_base", "pop_2010_ajustada", "pop_2022"]:
    base_demografica[col] = pd.to_numeric(base_demografica[col], errors="coerce")

base_demografica.describe().round(2)
```

```python
base_demografica[base_demografica["sigla_uf"].isna() | base_demografica["pop_2022"].isna()]
```

```python
resumo_estadual = base_demografica.groupby("sigla_uf", as_index=False)[["pop_2010_ajustada", "pop_2022"]].sum()
resumo_estadual
```

```python
resumo_estadual["variacao_populacional"] = resumo_estadual["pop_2022"] - resumo_estadual["pop_2010_ajustada"]
resumo_estadual["crescimento_percentual"] = (resumo_estadual["variacao_populacional"] / resumo_estadual["pop_2010_ajustada"] * 100).round(2)

resumo_estadual
```

```python
resumo_estadual = resumo_estadual.sort_values("variacao_populacional", ascending=False).reset_index(drop=True)
resumo_estadual
```

```python
resumo_estadual.to_csv("resumo_estadual_2010_2022.csv", index=False, encoding="utf-8-sig")
print("salvo: resumo_estadual_2010_2022.csv")
```

```python
dados_municipais = base_demografica[["sigla_uf", "codigo_uf", "codigo_mun", "nome_municipio", "pop_2010_ajustada", "pop_2022"]].copy()
dados_municipais
```

```python
dados_municipais["variacao_populacional"] = dados_municipais["pop_2022"] - dados_municipais["pop_2010_ajustada"]
dados_municipais["crescimento_percentual"] = (dados_municipais["variacao_populacional"] / dados_municipais["pop_2010_ajustada"] * 100).round(2)

dados_municipais
```

```python
dados_municipais = dados_municipais.sort_values("variacao_populacional", ascending=False).reset_index(drop=True)
dados_municipais.head(20)
```
